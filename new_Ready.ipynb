```python
import pandas as pd

df = pd.read_csv(r'D:\Ai project\adult 3.csv')
df.head()
```

```python
# Remove low education rows
df = df[~df['education'].isin(['1st-4th', '5th-6th', 'Preschool'])]

# Remove low education number (<= 4)
df = df[df['educational-num'] > 5]

# Remove uncommon occupations
df = df[df['occupation'] != 'Armed-Forces']

# Remove very old age entries (e.g., retired)
df = df[df['age'] <= 68]
```

```python
# Replace '?' with NaN
df.replace('?', pd.NA, inplace=True)

# Drop rows with missing values
df.dropna(inplace=True)
```

```python
# Remove raw 'education' since 'education-num' is numeric and more useful
df.drop('education', axis=1, inplace=True)
```

```python
# Age bucket (young, mid, senior)
df['age_group'] = pd.cut(df['age'], bins=[17, 30, 45, 68], labels=['Young', 'Mid', 'Senior'])

# Is highly educated
df['is_educated'] = (df['educational-num'] > 12).astype(int)

# Convert boolean to int
df['is_educated'] = df['is_educated'].astype(int)

df
```

```python
df = pd.get_dummies(df, drop_first=True)
clean_file = df
clean_file
```

```python
# 📌 Step 1: Import Libraries
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score, classification_report
from imblearn.over_sampling import SMOTE

# 📌 Step 2: Load and Clean Data
# Assuming you already have the refined DataFrame `df`
print(df.columns.tolist())
df.dropna(inplace=True)
df.rename(columns={'income_>50K': 'income'}, inplace=True)

# 📌 Step 3: Split Features & Labels
X = df.drop('income', axis=1)
y = df['income']

# 📌 Step 4: Scale Features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 📌 Step 5: Balance Dataset with SMOTE
smote = SMOTE(random_state=42)
X_resampled, y_resampled = smote.fit_resample(X_scaled, y)

# 📌 Step 6: Train/Test Split
X_train, X_test, y_train, y_test = train_test_split(
    X_resampled, y_resampled, test_size=0.2, stratify=y_resampled, random_state=42
)

# 📌 Step 7: Train MLPClassifier with Tuned Parameters
model = MLPClassifier(
    hidden_layer_sizes=(16, 8, 4),
    activation='relu',
    solver='adam',
    learning_rate='adaptive',
    alpha=0.0001,
    early_stopping=True,
    max_iter=1000,
    random_state=42
)

model.fit(X_train, y_train)

# 📌 Step 8: Evaluation
y_pred = model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)

print("✅ Final Accuracy:", round(accuracy * 100, 2), "%")
print("\n🔍 Classification Report:\n", classification_report(y_test, y_pred))
```
